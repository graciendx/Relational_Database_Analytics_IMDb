# IMDb Catalogue Analytics: Interpretation Guide

**Grace Ngo**

The project connects three catalogue views through one question: how do the data model and eligibility choices shape an analytical conclusion?

## Runtime summaries

Each genre/bin result contains a mean title rating and a support count. The original bare-column query did not establish the averages described in the earlier report. Corrected summaries remain descriptive and cannot identify an optimal editing runtime.

## Genre coverage

Exact multi-label membership replaces substring matching. Counts describe the supplied catalogue, not viewing demand. Each share uses all eligible movies in the period, including those with unknown genres. Shares across genres may exceed 100%. Partial decade windows are labeled.

## Ranked shortlists

The year view orders eligible movies by rating, votes and title ID. Vote-floor sensitivity reports candidate counts and overlap with the lowest-floor shortlist. No first-listed genre is called a primary genre, and votes are not equated with viewership.

## Evidence status

The original `imdb.db` was not supplied. Published results use explicitly fictional verification records. They establish runnable behavior, not new IMDb findings. A documented local-database rerun is the next empirical milestone.

## Technical delivery

The repository contains parameterized SQL, read-only database access, a streaming local TSV importer, sixteen behavioral and numerical tests, source fingerprints and an automated demonstration run. See README.md and docs/methodology.md for details.
